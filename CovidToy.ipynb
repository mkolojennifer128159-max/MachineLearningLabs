```python
import pandas as pd
from google.colab import files

uploaded = files.upload()

filename = next(iter(uploaded))
df = pd.read_csv(filename)

df.head()
```

```python
df.info()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder, LabelEncoder
```

```python
transformer = [
    ('trans1', SimpleImputer(), ['fever']),

    ('trans2', OrdinalEncoder(categories=[['Mild', 'Strong']]), ['cough']),

    ('trans3', OneHotEncoder(sparse_output=False, drop='first'), ['gender', 'city'])
]
```

```python
ct = ColumnTransformer(
    transformers=transformer,
    remainder='passthrough'
)

X = df.drop(columns=['has_covid'])

X_transformed = ct.fit_transform(X)

print(X_transformed)
```

```python
X_transformed_df = pd.DataFrame(
    X_transformed,
    columns=ct.get_feature_names_out()
)

X_transformed_df.head()
```

```python
le = LabelEncoder()

y = le.fit_transform(df['has_covid'])

print("Original classes:", le.classes_)
print("Encoded target:", y)
```

```python
final_df = X_transformed_df.copy()

final_df['has_covid'] = y

final_df.head()
```
